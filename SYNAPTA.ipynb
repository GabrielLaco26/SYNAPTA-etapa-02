## Preço médio e preço mediano



> Calcular o preço médio e o preço mediano dos aluguéis e explicar, de forma simples, por que os dois
valores podem ser diferentes.

Vamos pegar o valor médio e o valor mediano dos alugueis que estão presentes no arquivo e comparar.

Qual a diferença?


*   **Média:** Soma de todos os valores de aluguel dividido pela quantidade de apartamentos.
*   **Mediana:** Valor que está exatamente no meio, onde uma metade custa menos e a outra custa mais.

A diferença acontece pois na média pode ter um valor alto onde acaba aumentando a média geral, já na mediana temos o valor que está exatamente no centro de todos os outros.

Na lista abaixo teremos as partes separadas do código e o que cada uma faz:

* **#1** - Criamos uma variável para ter acesso a coluna aluguel.

* **#2** - Criamos a variável 'media', acessamos a coluna, usamos '.mean()' e o python faz o resto.

* **#3** - Criamos a variável mediana, acessamos a coluna, usamos '.median()' e o python faz o resto.




````python
# 1. Criar variável
coluna_aluguel = tabela["rent"]

# 2. Calcular a média
media = coluna_aluguel.mean()
print(media.round(1))

# 3. Calcular a mediana
mediana = coluna_aluguel.median()
print(mediana)
````

## Localidades com maiores e menores preços médios


>Identificar as localidades com maiores e menores preços médios, utilizando apenas grupos com quantidade razoável de registros.


Nesta etapa, vamos agrupar os imóveis por bairro/localidade para descobrir onde o aluguel é mais caro e onde é mais acessível.

**Por que filtrar por uma quantidade mínima de registros?**
Se um bairro tiver apenas 1 ou 2 imóveis anunciados e um deles for um imóvel de altíssimo padrão (outlier), a média desse bairro será distorcida. Para evitar análises enganosas, consideramos apenas bairros que possuem pelo menos **10 imóveis** cadastrados.

Na lista abaixo teremos as partes do código e o que cada uma faz:

* **#1** - Agrupamento e Métricas: Agrupamos a tabela pela coluna da localidade district, calculamos a média do aluguel 'rent' e contamos a quantidade de imóveis 'count'.
* **#2** - Filtro de Segurança: Filtramos apenas os bairros que têm 10 ou mais imóveis cadastrados.
* **#3** - Maiores Preços: Ordenamos os dados do maior para o menor preço médio para ver as áreas mais caras.
* **#4** - Menores Preços: Ordenamos os dados do menor para o maior preço médio para ver as áreas mais acessíveis.

**OBS:** O código `.groupby()` agrupa as categorias.

````python
# 1. Agrupar por distrito/localidade calculando preço médio e quantidade de imóveis
resumo_localidades = tabela.groupby('district').agg(
    preco_medio=('rent', 'mean'),
    quantidade_imoveis=('rent', 'count')
).reset_index()

# 2. Filtrar apenas grupos com quantidade razoável (mínimo de 10 registos)
limite_minimo = 10
locais_filtrados = resumo_localidades[resumo_localidades['quantidade_imoveis'] >= limite_minimo]

# 3. Localidades com MAIORES preços médios
maiores_precos = locais_filtrados.sort_values(by='preco_medio', ascending=False)
print("--- Top 5 Localidades com MAIORES Preços Médios ---")
display(maiores_precos.head(5).round(2))

# 4. Localidades com MENORES preços médios
menores_precos = locais_filtrados.sort_values(by='preco_medio', ascending=True)
print("\n--- Top 5 Localidades com MENORES Preços Médios ---")
display(menores_precos.head(5).round(2))
````

## Comparando o valor do aluguel com características do imóvel


> Comparar o valor do aluguel com pelo menos duas características disponíveis, como área, quantidade de quartos, banheiros ou vagas.


Agora vamos analisar como o valor do aluguel se comporta de acordo com características físicas dos imóveis, especificamente a quantidade de quartos e o espaço da garagem.

O objetivo é observar se o aumento dessas facilidades acompanha uma elevação proporcional no preço do aluguel.

Na lista abaixo teremos as partes do código e o que cada uma faz:
* **#1 - Análise por Quartos:** Agrupamos os dados por número de quartos para ver o preço médio de cada categoria.
* **#2 - Análise por Garagem:** Agrupamos os dados por número de garagens para comparar o impacto dessa característica no valor.

````python
# 1. Agrupar por quantidade de quartos e calcular o preço médio e o total de imóveis
analise_quartos = tabela.groupby('bedrooms').agg(
    preco_medio=('rent', 'mean'),
    total_imoveis=('rent', 'count')
).reset_index()

print("--- Preço Médio por Quantidade de Quartos ---")
display(analise_quartos.sort_values(by='bedrooms').round(2))

# 2. Agrupar por quantidade de garagens/vagas e calcular o preço médio e o total de imóveis
analise_garagem = tabela.groupby('garage').agg(
    preco_medio=('rent', 'mean'),
    total_imoveis=('rent', 'count')
).reset_index()

print("\n--- Preço Médio por Vagas de Garagem ---")
display(analise_garagem.sort_values(by='garage').round(2))
````

## Gráfico: Preço médio do aluguel por tipo de imóvel

> Criar pelo menos três gráficos simples e legíveis, com título, identificação dos eixos e breve explicação do que cada gráfico mostra.

No próximo bloco de código iremos:

1. Agrupar os dados por tipo de imóvel (`type`).
2. Calcular o preço médio do aluguel (`rent`) para cada categoria.
3. Ordenar os valores para facilitar a comparação visual.
4. Construir um gráfico de colunas com título, eixos identificados e rótulos legíveis.

Para isso, utilizaremos as bibliotecas "pandas" para o tratamento dos dados e "matplotlib.pyplot" para a criação da visualização gráfica.

Na lista abaixo teremos as partes separadas do código e o que cada uma faz:

- **#1** - Agrupamos a tabela por tipo de imóvel (`type`) e calculamos a média (`mean`) do valor do aluguel (`rent`).
- **#2** - Ordenamos os dados em ordem decrescente (`ascending=False`) para destacar quais tipos de imóveis possuem os maiores preços médios.
- **#3** - Configuramos a estrutura do gráfico de colunas (`plt.bar`), definindo o título principal e nomeando os eixos X (tipo de imóvel) e Y (preço médio em R$).
- **#4** - Ajustamos a rotação dos rótulos do eixo X para evitar sobreposição de texto e exibimos o gráfico final com `plt.show()`.

**OBS:** O código `plt.xlabel` / `plt.ylabel` servem para identificar os eixos do gráfico.

````python
import matplotlib.pyplot as plt

# 1. Agrupar por tipo de imóvel e calcular o preço médio do aluguel
preco_por_tipo = tabela.groupby('type')['rent'].mean().reset_index()

# 2. Ordenar os valores do maior para o menor preço médio
preco_por_tipo = preco_por_tipo.sort_values(by='rent', ascending=False)

# 3. Criar o gráfico de colunas
plt.figure(figsize=(10, 6))
plt.bar(preco_por_tipo['type'], preco_por_tipo['rent'], color='royalblue')

# Identificação dos eixos, título e melhorias de legibilidade
plt.title('Comparação do Preço Médio de Aluguel por Tipo de Imóvel', fontsize=14, fontweight='bold')
plt.xlabel('Tipo de Imóvel', fontsize=12)
plt.ylabel('Preço Médio do Aluguel (R$)', fontsize=12)
plt.xticks(rotation=45, ha='right')

# Ajustar layout para que nenhum texto seja cortado
plt.tight_layout()

# 4. Exibir o gráfico na tela
plt.show()
````

## Gráfico: Distribuição da oferta de aluguéis por quantidade de quartos

> Criar pelo menos três gráficos simples e legíveis, com título, identificação dos eixos e breve explicação do que cada gráfico mostra.

No próximo bloco de código iremos:

1. Agrupar os dados por número de quartos (`bedrooms`).
2. Contar a quantidade total de imóveis disponíveis em cada categoria.
3. Ordenar os dados sequencialmente pela quantidade de quartos.
4. Construir um gráfico de colunas com título, eixos identificados e visualização limpa.

Na lista abaixo teremos as partes separadas do código e o que cada uma faz:

- **#1** - Agrupamos a tabela por quantidade de quartos (`bedrooms`) e contamos quantos imóveis existem em cada categoria usando `.count()`.
- **#2** - Renomeamos a coluna contada e ordenamos os registros em ordem crescente (`ascending=True`) para que o eixo X siga uma progressão lógica (1, 2, 3 quartos, etc.).
- **#3** - Configuramos a estrutura do gráfico de colunas (`plt.bar`), definindo o título principal e nomeando os eixos X (quantidade de quartos) e Y (quantidade total de imóveis).
- **#4** - Ajustamos o layout para garantir legibilidade e exibimos o gráfico final na tela com `plt.show()`.

````python
import matplotlib.pyplot as plt

# 1. Agrupar por quantidade de quartos e contar o total de imóveis
imoveis_por_quarto = tabela.groupby('bedrooms')['rent'].count().reset_index()
imoveis_por_quarto.columns = ['bedrooms', 'quantidade']

# 2. Ordenar os valores pela quantidade de quartos em ordem crescente
imoveis_por_quarto = imoveis_por_quarto.sort_values(by='bedrooms', ascending=True)

# 3. Criar o gráfico de colunas
plt.figure(figsize=(10, 6))
plt.bar(imoveis_por_quarto['bedrooms'].astype(str), imoveis_por_quarto['quantidade'], color='teal')

# Identificação dos eixos, título e formatação
plt.title('Distribuição da Oferta de Aluguéis por Quantidade de Quartos', fontsize=14, fontweight='bold')
plt.xlabel('Quantidade de Quartos', fontsize=12)
plt.ylabel('Quantidade de Imóveis', fontsize=12)

plt.tight_layout()

# 4. Exibir o gráfico na tela
plt.show()
````

Na lista abaixo teremos as partes separadas do código e o que cada uma faz:

- **#1** - Agrupamos a tabela por quantidade de vagas de garagem (`garage`) e calculamos a média (`mean`) do valor do aluguel (`rent`).
- **#2** - Ordenamos os dados em ordem crescente (`ascending=True`) com base na quantidade de vagas para manter a progressão lógica no eixo X.
- **#3** - Configuramos a estrutura do gráfico de colunas (`plt.bar`), definindo o título principal e nomeando os eixos X (quantidade de vagas) e Y (preço médio do aluguel em R$).
